# Tutorium Programmieren (Prof. Dr. Ralf Gerlich) - Aufgabenblatt 9

Implementieren Sie Unit-Tests mit `pytest` für die Methoden
- `Person.nächster_geburtstag`,
- `Person.alter`,
- `Personendatenbank.einfügen`,
- `Personendatenbank.findePerson` und
- `Personendatenkbank.entfernen`.

Überlegen Sie sich anhand Ihres Codes, welche unterschiedlichen Fälle relevant sein könnten und implementieren Sie die entsprechenden Tests. Testen Sie Ihr Modul mit diesen Tests.

Messen Sie die Coverage mit `pytest-cov` ([Installationshinweise](https://pytest-cov.readthedocs.io/en/latest/readme.html#installation), [Nutzungshinweise](https://pytest-cov.readthedocs.io/en/latest/readme.html#usage)).

**Hinweis**: Implementieren Sie Ihre Lösung in PyCharm (auf Basis der Lösung des vorigen Aufgabenblatts) und reichen Sie eine ZIP-Datei mit Ihrem Projektverzeichnis in FELIX ein.

**Fragen**:
1. Welche Schwierigkeiten könnten sich ergeben, wenn wir auch die Funktion `laden` und die Methode `speichern` mit `pytest` testen wollten?
2. Könnten wir auch das Hauptprogramm mit `pytest` testen?
