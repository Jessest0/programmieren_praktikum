# Tutorium Programmieren (Prof. Dr. Ralf Gerlich) - Aufgabenblatt 7

Ergänzen bzw. modifizieren Sie das Modul `personendatenbank` aus dem vorigen Aufgabenblatt wie folgt:
1. Implementieren Sie eine Klasse `Name` mit den Feldern `name` und `vorname` und einem geeigneten Konstruktor.
2. Implementieren Sie die erforderlichen Methoden, so dass Objekte der Klasse `Name` als Schlüssel für Dictionaries verwendet werden können, und zwei Objekte dann als gleich erkannt werden, wenn die Felder `name` und `vorname` jeweils den gleichen Inhalt haben.
3. Verändern Sie die Klasse `Person` so, dass statt den Feldern `name` und `vorname` nur noch ein Feld `name` vorhanden ist, in dem ein Objekt der neuen Klasse `Name` gespeichert wird. Passen Sie den Rest Ihres Codes an diese Änderung an (z.B. die Funktion `laden` und die Methode `speichern`).
4. Stellen Sie die Klasse `Personendatenbank` so um, dass die enthaltenen `Personen`-Objekte nun in einem Dictionary gespeichert werden, dessen Schlüssel Objekte vom Typ `Name` und die Werte Objekte vom Typ `Person` sind. Passen Sie den Rest Ihres Codes an diese Änderung an (z.B. die Funktionen `findePerson` oder `entferne`).

**Wichtig!** Testen Sie nach jedem dieser Schritte Ihr Programm, um festzustellen, ob es noch funktioniert! Machen Sie nach jedem Schritt eine Sicherungskopie Ihres Codes, damit Sie zu einer funktionierenden Version zurückgehen können, sollten Sie einen Fehler im Programm eingebaut haben, den Sie nicht mehr nachvollziehen können!
